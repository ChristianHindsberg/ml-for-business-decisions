# Airline tweets: loading the data

Loads the Twitter US Airline Sentiment data and previews it. The dataset is not in this repo (see `data/README.md`).

```python
import pandas as pd
import sys
from pathlib import Path

# Data lives in the repo's data/ folder (see data/README.md).
# In Colab, upload the file to /content/data/ or mount Drive and point DATA_DIR at it.
IN_COLAB = 'google.colab' in sys.modules
DATA_DIR = Path('/content/data') if IN_COLAB else Path('../data')

file_path = DATA_DIR / 'Airline_Tweets.csv'

try:
    df_twitter = pd.read_csv(file_path)
    print(f"Successfully loaded data from {file_path}")
    display(df_twitter.head())
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found. Please ensure the path is correct and your Google Drive is mounted correctly.")
except Exception as e:
    print(f"An error occurred: {e}")
```
